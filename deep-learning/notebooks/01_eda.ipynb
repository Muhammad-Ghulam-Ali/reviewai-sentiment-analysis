```python
# Libraries
import sys
sys.path.append('..')

from src.utils import section

import pandas as pd
```

```python
# Load dataset
from src.data_load import data_loader

df = data_loader('../data/raw/review_train.csv')
df.head()
```

```python
# Eda
section('shape', 45)
print(f"Rows:             {df.shape[0]:,}\nColumns:          {df.shape[1]}")

section('columns and dtypes', 45)
display(df.dtypes.to_frame('dtype'))

section('head', 45)
display(df.head())

section('nulls', 45)
display(df.isnull().sum().to_frame('nulls'))

section('duplicates', 45)
print(f"Duplicate rows:   {df.duplicated().sum():,}")
print(f"Duplicate texts:  {df['text'].duplicated().sum():,}")

section('label distribution', 45)
counts = df['label'].value_counts().sort_index()
display(pd.DataFrame({'count': counts, 'percent': (counts / len(df) * 100).round(2)}))

section('review length (words)', 45)
df['word_count'] = df['text'].str.split().str.len()
display(df['word_count'].describe().round(1).to_frame('words'))
print(f"90th percentile:  {df['word_count'].quantile(0.90):,.0f}")
print(f"95th percentile:  {df['word_count'].quantile(0.95):,.0f}")
print(f"99th percentile:  {df['word_count'].quantile(0.99):,.0f}")

section('very short reviews', 45)
print(f"Under 3 words:    {(df['word_count'] < 3).sum():,}")
```
