```python
# Libraries
import sys
sys.path.append('..')
import joblib

from src.utils import section
```

```python
# Load data
from src.data_load import data_loader

df = data_loader('../data/processed/clean_train_review.csv')
test_df = data_loader('../data/processed/clean_test_review.csv')

df.head()
```

```python
# Split the data
from src.text_processing import data_splitter

X_train, X_val, y_train, y_val = data_splitter(df['text'], df['label'])

section("Shape of data", 30)
print(f"""
X_train:             {X_train.shape[0]:,}
X_val:               {X_val.shape[0]:,}
""")
```

```python
# Text preprocess

from src.text_processing import build_vocab, encode

vocab = build_vocab(X_train)

X_train_id = encode(X_train, vocab)
X_val_id = encode(X_val, vocab)
```

```python
# Pad
from src.text_processing import pad_id

X_train_pad = pad_id(X_train_id)
X_val_pad = pad_id(X_val_id)
```

```python
# Test data
from src.preprocessing import make_binary

df_test = make_binary(data_loader("../data/raw/review_test.csv"))
X_test, y_test = df_test['text'], df_test['label']

X_test_id = encode(X_test, vocab)
X_test_pad = pad_id(X_test_id)
```

```python
# Save
items = {
    "X_train_pad": X_train_pad,
    "X_val_pad": X_val_pad,
    "X_test_pad": X_test_pad,
    "y_train": y_train,
    "y_val": y_val,
    "y_test": y_test,
    "vocab": vocab,
}

for name, obj in items.items():
    joblib.dump(obj, f"../data/processed/{name}.pkl")
```
